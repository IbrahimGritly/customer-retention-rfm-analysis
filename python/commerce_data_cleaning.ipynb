# Find missing rows

```python
# Count unique entries in 'Customer ID' column
df['Customer ID'].nunique()
print(f"There are {df['Customer ID'].nunique()} unique customer IDs.")

df['Customer ID'].isnull().sum() # Check how many rows in Customer ID are missing
print(f"There are {df['Customer ID'].isnull().sum()} missing rows customer IDs.")
```

```python
df = df.dropna(subset=["Customer ID"]) # Drop rows with missing Customer ID
```

```python
df.duplicated().sum() # Check for duplicate rows
```

```python
# Find how many rows in Customer ID are missing after we dropped them above
df['Customer ID'].isnull().sum()
```

# Remove negative or 0 quantities and prices

```python
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)]
```

```python
df.isna().sum() # No more missing values
```

# Fix the date format

```python
# Check data type of InvoiceDate column
print(df['InvoiceDate'].dtype)
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
print(f"The new data type for InvoiceDate is {df['InvoiceDate'].dtype}.")
```

```python
df['Customer ID'].isnull().sum()
```

# Remove duplicates
Exact duplicate rows were identified and removed to prevent inflated revenue, order counts, and customer activity metrics. These duplicates represent data entry or system replication errors rather than legitimate transactions.

```python
df.duplicated().sum() 
print(f"There are {df.duplicated().sum()} duplicate rows remaining in the dataset.")
```

```python
df.drop_duplicates(inplace=True) # Drop duplicate rows
```

```python
print(f"There are {df.duplicated().sum()} duplicate rows remaining in the dataset.")
```
